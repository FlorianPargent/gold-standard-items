This article investigates whether gold-standard items are helpful for questionnaire development. Based on an initial item pool (138 items) derived through a deductive item construction ($N$ = 124) and a prototype approach ($N$ = 24), followed by cognitive interviews ($N$ = 8) and a pilot study ($N$ = 390) of a preliminary item set (61 items), three 12-item scales were constructed to measure the physical traits of body height, body weight, and age. We collected data on these scales using response formats with either two ($N$ = 921) or six ($N$ = 933) categories. We also collected numeric self-reports of body weight, body height, and age as gold-standard items. We argue that the items intended to measure the latent variable (i.e., body height, body weight, age) most directly (i.e., the gold-standard items) should have the highest loadings on that variable. Contrary to this assumption, confirmatory factor analyses revealed that the gold-standard items did not consistently exhibit the highest loadings on their corresponding latent variables, calling into question the content validity of our scales. Furthermore, when controlling for the self-reported physical body height, body weight, and age as gold-standard items, as well as gender, we did not always find an interpretable, systematic residual variance. Finally, the pattern of correlations among the latent variables did not closely reflect those among the self-reported gold-standard items, suggesting that the item scales and the gold-standard items do not share the same validity. While these results are consistent with previous studies, our analyses also showed that items with two response categories were as valid as those with six categories, contradicting past recommendations. Our analyses underscore the importance of intensive cognitive pretesting in scale construction and suggest that questionnaires using different response formats should be compared empirically and that hypotheses about which items best represent the latent variable should be tested.

When constructing items for a new scale, the common approach is to first carefully define the construct or latent variable to be measured, and then generate indicators for it based on this definition (see [Anastasi & Urbina, 1997](#ref-anastasiPsychologicalTests1997)). When based on common factor analytic test models in which each item is an indicator of only one latent variable, items are theoretically interchangeable with respect to their latent variable; that means they all measure the same construct. In theory, the item selection is considered content valid if the items represent a random selection of all possible items that could be used to measure the latent variable. However, in many cases, the item pool is unknown, so it is not possible to select items at random. Thus, in practice, when items are not sampled randomly, a scale is considered content valid only if each aspect of the construct definition is reflected by at least one item. The items of a scale are then used together to measure the latent variable. When evaluating such a scale empirically, an important goal is to assess whether the questionnaire actually measures the intended construct. While standard practices report model fit indices and evaluate whether each item has a substantial response variance and item loading, this does not directly answer the question of what the items actually measure. And even if researchers go the extra mile and reason about the items by interpreting the observed pattern of item loadings (i.e., which items have relatively higher loadings), they often lack a clear strategy for what to compare them to and which patterns would suggest that the items measure the construct as intended. In the most widely applied test-theoretic models (e.g., factor analytic models), item loadings are interpreted as the extent to which items correlate with, and thus measure, their latent variable (see [Lord et al., 1968](#ref-lordStatisticalTheoriesMental1968)). Would it not be great if we had at least one indicator with a strong theoretical justification that this *gold-standard item* is a particularly good measure of the latent variable? Then we could include the gold-standard item as an additional indicator in the questionnaire’s evaluation and check whether it has the highest, or at least a very high, loading among the items used to measure the latent variable. If other items load substantially higher than the gold-standard item, one might rightly question whether the scale measures the construct as originally intended, and could start investigating the actual meaning of the latent variable by making sense of the complete loading pattern.

In this study, we examine the idea of evaluating psychological scales with gold-standard items by constructing three psychological questionnaires for the physical traits of *body height*, *body weight*, and *chronological age*. Extending previous examples of psychological measures of body height ([Bortolotti et al., 2013](#ref-bortolottiRelevanceAdvantagesUsing2013); [Hilbert et al., 2022](#ref-hilbertWhatMeasureEmpirical2022); [Van der Linden, 2016](#ref-vanderlindenUnidimensionalLogisticResponse2016)), we chose these physical traits because, unlike psychological latent variables, they can be directly assessed using valid gold-standard items in the form of numerical self-reports of the respective physical trait (e.g., self-reported body height in centimeters). We derive and evaluate three assumptions that we argue should hold in our idealized setting with theoretically strong gold-standard items, assuming that our psychological scales achieved valid measurements of the respective latent variables. To reduce the risk of confounding our findings by how we constructed our psychological scales, we adhere to best practices in the literature on scale construction and item wording. We also compare questionnaire versions with two different response formats as an additional robustness check. After presenting our analyses, we discuss the implementation of gold-standard items for evaluating psychological scales for more common psychological constructs. Furthermore, we discuss the overlap between gold-standard items and single-item measures, which are becoming increasingly common in applied psychological research ([Allen et al., 2022](#ref-allenSingleItem2022)).

## Theoretical Background

### Gold-Standard Items for Physical Traits

The idea to construct psychological scales for physical traits is not new: To illustrate the usefulness of logistic item response models, Van der Linden ([2016](#ref-vanderlindenUnidimensionalLogisticResponse2016)) analyzed a teaching dataset from the Netherlands in which body height was measured based on items with two response categories (e.g., “I bump my head quite often.”, p. 27). He notes that this approach is unusual, as body height appears to be a physical variable that can only be measured using a yardstick. However, he argues that because differences in height do influence behavior, it is plausible to treat such behavioral indicators as proxies for body height and use them to construct a psychological scale of height. As the author himself noted, a limitation of his dataset was that it did not include a physical height measurement for comparison.

In contrast, Bortolotti et al. ([2013](#ref-bortolottiRelevanceAdvantagesUsing2013)) compared psychological and physical measurements to demonstrate the validity of psychological measurements and the practical utility of psychological measurement models. They also constructed a 27-item scale to assess body height using a binary response format. Participants from Brazil were asked to respond to items such as “Do I think I would do well in a basketball team?” ([Bortolotti et al., 2013, p. 2350](#ref-bortolottiRelevanceAdvantagesUsing2013)). At the end of the scale, participants also indicated their actual height in centimeters, a self-reported physical gold-standard item. Bortolotti et al. compared the parameters estimated from the latent variable body height with the response on the gold-standard item. Their analyses revealed a strong correlation of .86 between self-reported physical height and the estimated person parameter derived from a two-parameter logistic model of the psychological scale. However, the reported analyses do not establish whether the self-reported physical height item would have had the highest standardized loading in a model including both this item and the questionnaire items ([Bortolotti et al., 2013](#ref-bortolottiRelevanceAdvantagesUsing2013)).

Although both studies concluded that their scales successfully measured physical body height, they also identified some items that did not fit the one-dimensional model well, had extreme item difficulties, or did not have a linear relationship with self-reported physical height. These findings show that even for a relatively simple trait such as physical body height, it is not trivial to construct appropriate items. In particular, the examples of item misfit in Van der Linden ([2016](#ref-vanderlindenUnidimensionalLogisticResponse2016)) suggest that, in addition to physical height, the items also measured a psychological representation of body height, but the author did not further investigate possible interpretations for this residual psychological component.

### Item Wording

One important aspect of item construction is item wording. Surprisingly, Pargent et al. ([2019](#ref-pargentCanMakeIt2019)) and Goretzko et al. ([2020](#ref-goretzkoNotVeryPowerful2020)) showed that the quality of item wording did not, in fact, impact model fit based on commonly used fit indices in confirmatory factor analysis. However, clear wording is indisputably essential for understanding an item correctly, highlighting the boundaries of fit indices and emphasizing the importance of conducting cognitive interviews with individuals from the target population when developing items for a new scale ([Gehlbach & Brinkworth, 2011](#ref-gehlbachMeasureTwice2011); [Willis, 2005](#ref-willisCognitiveInterviewing2005); [Willis & Artino, 2013](#ref-willisWhatDo2013)). If a person does not understand an item correctly, their response can never be valid, regardless of any model fit indices (for an extreme but illustrative example, see [Maul, 2017](#ref-maulRethinking2017)). Thus, it seems problematic to evaluate the quality of a questionnaire solely by looking at the fit of the latent variable model, item loadings, or reliability estimates. In this context, McNeish et al. ([2018](#ref-mcneishThornyRelationMeasurement2018)) stated that fit indices (e.g., Root Mean Square Error of Approximation, RMSEA; Standardized Root Mean Square Residual, SRMR; and the Comparative Fit Index, CFI) should always be considered in combination with measurement quality (i.e., reliability) and the size of standardized factor loadings. Gold-standard items offer an additional systematic strategy for scale validation by using the pattern of factor loadings to provide a reference point for evaluating the meaning of other items.

### Response Format

An important topic that hinders the generalization of methodological insights across questionnaires is the potential effect of response formats on scale quality. Hilbert et al. ([2016](#ref-hilbertInfluenceResponseFormat2016)) presented initial evidence that different response formats might not measure the same latent variable. They found that the correlation between latent variables measured by identically worded items ranged from .76 to .88 when using scales with two versus five response categories, as well as a visual analogue scale. The question arises: Which of these questionnaires best measures the intended construct?

What we do know in this context is how the number of response categories affects reliability estimates, because a vast body of literature has compared different numbers of response categories for rating-scale items. We cite several exemplary studies to demonstrate how complex and difficult it is to choose an optimal number of response categories and to highlight what we can conclude from their results. First, Preston & Colman ([2000](#ref-prestonOptimalNumberResponse2000)) reported that two response categories had the lowest retest reliability and relatively poor validity and discriminative power. They recommended applying rating scales with at least seven response categories. In contrast, Lee & Paek ([2014, p. 663](#ref-leeSearchOptimalNumber2014)) concluded that “caution should be made if a scale has only two response categories, but that limitation may be overcome by manipulating other scale features, namely, scale length or item discrimination”. Weijters et al. ([2010](#ref-weijtersEffectRatingScale2010)) recommend rating scales with five fully labeled response categories for samples drawn from the general population. In their recommendations, the authors considered an extreme response style, responses to reversed items, and acquiescence. However, Masuda et al. ([2017](#ref-masudaRespondentsLowMotivation2017)) found that persons with low motivation tend to use the middle category, implying that a middle category is not advisable. Similarly, Bradley et al. ([2015, p. 8](#ref-bradleyRatingScalesSurvey2015)) concluded from their study that “for constructing measures from survey responses, the inclusion of a neutral middle category distorts the data to the point where it is not possible to construct meaningful measures.” Simms et al. ([2019](#ref-simmsDoesNumberResponse2019)) suggest that researchers concerned about the ambiguity of a middle category can use six response categories, as their study found no further gains in psychometric precision beyond six categories. All in all, the reviewed literature suggests that to achieve high reliability, seven response categories might be slightly favored, but there is also evidence to omit the middle category due to its inconsistent use.

## The Present Study

We quote from a review on the application of item response models and factor analysis ([Holt et al., 2010, p. 288](#ref-tenholtScaleConstructionEvaluation2010)): “In our opinion, researchers could take far better advantage of their theoretical knowledge and/or expectations by incorporating their a priori knowledge of the items and scales in the analyses. This should be reflected 1) by a more frequent application of confirmatory techniques, especially in the construction of new scales, and 2) by adding interpretability of factors and item contents to the criteria used for model evaluation.” In line with this statement and the aforementioned research and recommendations, we assess three physical traits to replicate and extend previous findings. First, we constructed scales for body height, body weight, and age. Second, we used items with two response categories to replicate the results for body height reported by Bortolotti et al. ([2013](#ref-bortolottiRelevanceAdvantagesUsing2013)). Third, we applied an additional format with six response categories, omitting the middle category, which is often overused by respondents with low motivation (e.g., see [Masuda et al., 2017](#ref-masudaRespondentsLowMotivation2017)). We aimed to investigate whether the results of Bortolotti et al. ([2013](#ref-bortolottiRelevanceAdvantagesUsing2013)) hold true for body height and other physical traits, and whether we can identify a residual psychological component in self-report questionnaires of physical traits, as indicated by Van der Linden ([2016](#ref-vanderlindenUnidimensionalLogisticResponse2016)). Further, we investigate whether the main results are robust across response formats and whether the majority view in the literature is correct in assuming that items with six response categories have higher reliability and, possibly, better validity than items with two response categories ([Lee & Paek, 2014](#ref-leeSearchOptimalNumber2014); [Preston & Colman, 2000](#ref-prestonOptimalNumberResponse2000); [Weijters et al., 2010](#ref-weijtersEffectRatingScale2010)). In more detail, we argue that if our new scales measure their respective physical trait as intended, we expect to observe the following patterns:

- *Assumption 1:* The gold-standard items (self-reported physical body height, body weight, and age) should have the highest or at least one of the highest factor loadings among other questionnaire items measuring the respective physical trait (because in one-factor measurement models, a standardized loading can be interpreted as the correlation between the latent variable and the item).

- *Assumption 2:* Even if gold-standard items have the highest factor loading (and especially if *Assumption 1* does not hold), the measurement of physical traits with questionnaire items possibly captures an additional component that can be interpreted as a psychological representation of the physical trait. When item responses are predicted by the gold-standard item and gender (as we would expect gender differences in self-perception, see [McCreary, 2002](#ref-mccrearyGenderAgeDifferences2002)), there should remain systematic variance that can be explained by a common latent variable. Ideally, the loading pattern on this residual latent variable should be interpretable to identify the meaning of the psychological component.

- *Assumption 3a, b:* The correlation pattern among the latent variables of the physical traits (in a three-factor model with correlated latent variables) should closely correspond to the correlations among the gold-standard items (3a). For scales with six response categories, reliability estimates should be higher, and correlations among latent variables and among gold-standard items should be more closely aligned than for scales with two response categories (3b).

## Scale Construction

### Initial Item Pool

First, we constructed items for the physical traits of body height, body weight, and age. Age was defined as chronological age in years, corresponding to the time elapsed since a person’s birth ([Montepare & Lachman, 1989](#ref-montepareYouReOnly1989); [Schwall, 2012](#ref-schwallDefiningAgeUsing2012)). Height was defined as the height of an upright person from the sole of the foot to the top of the head in centimeters, and body weight as the physical mass of a person in kilograms ([Martin, 1929](#ref-martinAnthropometrieAnleitungSelbstaendigen1929)). The initial items were developed by 124 psychology students (through deductive item construction, [Burisch, 1984](#ref-burischApproachesPersonalityInventory1984)) and 24 persons (through a prototype approach, see [Broughton, 1984](#ref-broughtonPrototypeStrategyConstruction1984)) who deviated from the German population mean by at least one standard deviation in the relevant characteristics (separately for women and men). The 24 persons with extreme trait expressions were asked to think of prototypical behaviors for each construct to assess the whole spectrum of the latent variable. This process resulted in an initial set of 138 items (response categories were not yet tested here).

### Cognitive Interviews

In a second step, we examined how these items were interpreted and whether they were connected to additional concepts unrelated to the constructs using cognitive interviews. We conducted interviews lasting two to three hours and applied various cognitive procedures, including probing, paraphrasing, concurrent think-aloud, and retrospective think-aloud ([Prüfer & Rexroth, 1996](#ref-prueferVerfahrenEvaluationSurvey1996), [2005](#ref-prueferKognitiveInterviews2005)). The literature on cognitive interviews recommends five to 30 interviews but indicates that the most serious problems in item wording can already be identified with a small number of interviews ([Willis, 2005](#ref-willisCognitiveInterviewing2005); [Willis & Artino, 2013](#ref-willisWhatDo2013)). The interviews were conducted with 8 people (4 women and 4 men) aged between 21 and 77 (*M* = 44.50, *SD* = 20.78). The level of education ranged from a high school diploma to a university degree. After the cognitive interviews, misleading items were reformulated or eliminated ([Faulbaum et al., 2009](#ref-faulbaumWasIstGute2009)), resulting in 61 remaining items.

### Pilot Study

In a third step, these 61 items were pilot-tested online with a new sample of initially 456 people. However, we excluded 66 subjects who had not completed the questionnaire, resulting in a sample of 390 participants aged between 18 and 77 years (*M* = 31.91, *SD* = 12.82), including 302 women (77.44%) and 88 men (22.56%). The level of education was distributed as follows: 0% no school-leaving certificate, 3.33% secondary school-leaving certificate/elementary school or equivalent, 17.18% secondary school or equivalent, 35.90% vocational baccalaureate or high school diploma, 40.77% college degree or university degree, 2.82% doctorate or habilitation. Of all respondents, 95.38% stated German as their native language. Based on these survey responses, we excluded items if a) their highest loading was not on the intended factor in three-dimensional exploratory factor analyses, b) their main and secondary loadings were nearly equal, or c) their main loading was less than or equal to .30. Additionally, if (d) items had very similar content, items with lower loadings were excluded. More details on the item construction process can be found in Eichhorn ([2019](#ref-eichhornItem2019)). We recomputed the exploratory factor analyses, item total correlations, and reliability estimates that were computed in this final item selection stage and report them in our electronic supplementary materials (ESM C).

No power analysis was performed to determine the sample size of the pilot study. Instead, the goal was to collect data from approximately 150 subjects. No data was analyzed before data collection ended.

### Final Scales

Based on these criteria, 12 final items were selected for each physical trait (see ([**apx-1?**](#ref-apx-1))).[1] For each of the three questionnaires, we created two versions for our main study – one with two response categories combining verbal labels (“does not apply” and “applies”) with numeric anchors (numbers 0 and 1) and one with six response categories combining verbal endpoints (“does not apply” to “applies”, only endpoints were labeled) with full numeric anchoring (numbers from 1 to 6). The two-category scale version was included because we wanted to replicate the results by Bortolotti et al. ([2013](#ref-bortolottiRelevanceAdvantagesUsing2013)) and Van der Linden ([2016](#ref-vanderlindenUnidimensionalLogisticResponse2016)), who also used a binary response format, and because this response format is commonly used in clinical assessment ([Clark & Watson, 2019](#ref-clarkConstructingValidityNew2019); [Rosellini & Brown, 2021](#ref-roselliniDevelopingValidatingClinical2021)). The six-category scale version was used as a robustness check to ensure that our results hold for ordinal response formats, which are common in practice, and to replicate the common finding that reliability estimates are higher for scales with more than two response categories ([Preston & Colman, 2000](#ref-prestonOptimalNumberResponse2000)). Using six response categories is an obvious synthesis of the literature, favoring a higher number of response categories ([Preston & Colman, 2000](#ref-prestonOptimalNumberResponse2000)) whilst avoiding a middle category ([Masuda et al., 2017](#ref-masudaRespondentsLowMotivation2017)), and there is also empirical support that six categories are a good choice ([Simms et al., 2019](#ref-simmsDoesNumberResponse2019)). However, we acknowledge that the evidence is not clear-cut, and other researchers recommend five or seven response categories, depending on the study population and purpose ([Weijters et al., 2010](#ref-weijtersEffectRatingScale2010)). We decided against including additional questionnaires with other response formats to ensure a sufficient sample size for each version.

## Statistical Analysis

All analyses for our main study were conducted in R version 4.5.3 (2026-03-11). A full list of software versions can be found in ([**apx-3?**](#ref-apx-3)). The data were analyzed using confirmatory factor analysis with the package *lavaan* ([Rosseel, 2012](#ref-R-lavaan)). We used the weighted least squares, mean and variance adjusted (WLSMV) estimator, treating the questionnaire items as ordinal and the self-reported physical gold-standard items body height, body weight, and age as continuous variables. First, we specified one-factor models with no error correlations and with correlated errors for each trait. In both models, the gold-standard items and the other 12 items were specified as indicators of a single latent variable. Second, to test whether an interpretable latent variable remains after controlling for the gold-standard item and gender, one-factor models were specified with a single latent variable, with the questionnaire items additionally predicted by the gold-standard item and gender.[2] Third, to compare the correlations between the latent variables of the physical traits with the correlations between the gold-standard items, one full three-factor model without correlated errors was estimated, in which only the questionnaire items load on their respective latent variable, and correlations are estimated between the latent variables, between the gold-standard items, and between the latent variables and the gold-standard items. We also conducted reliability analyses using the package *MBESS* ([Kelley, 2023](#ref-R-MBESS)) and report McDonald’s Omega as the reliability estimate of internal consistency for each scale. All analyses were run separately for both response formats (two-category and six-category scales).

The analyses reported in the current manuscript were not preregistered. However, the data we used have previously been analyzed in the first author’s German dissertation ([Eichhorn, 2019](#ref-eichhornItem2019)). As part of this dissertation, a preregistration was published at <https://osf.io/cz3uv/?view_only=9cac02db231b48629fea9ae53c3038b9>. The preregistration describes the planned data collection procedures and sample sizes, specifies the scale construction process, including the cognitive interviews and item selection based on the pilot sample, and registers the design of the questionnaires used in the validation sample, including the response format of both versions. In the current manuscript, we performed all of these steps as preregistered. However, our statistical analyses deviated from those listed in the preregistration. The reason for this deviation is that our general understanding of what we can learn from scales of physical traits about the construction of psychological scales has evolved continuously since preregistration, and we deem the analyses in the current manuscript more elegant for presenting our ideas than the originally preregistered analyses. In fact, the preregistration did not include the concept of gold-standard items or the detailed assumptions discussed in this manuscript. Instead, the registered research question was kept vague: *“The main research question is how well the items represent the real score values. More precisely, is it possible to construct a questionnaire, following the best practice of test construction, to measure a variable that we claim to measure.”* The results of the preregistered analyses are fully reported in Eichhorn ([2019](#ref-eichhornItem2019)). Note that no results in this manuscript are in conflict with any of the preregistered analyses. Both the current manuscript and the preregistration state that all analyses should be considered exploratory.

## Participants

The total sample of our main study consisted of 1854 participants aged between 18 and 97 years (*M* = 32.70, *SD* = 16.01), including 1036 women (56%) and 818 men (44%). The overall sample consisted of two independent subsamples, which were divided into the conditions with two ($N$ = 921) and six ($N$ = 933) response categories.

Participants were recruited in trains, colleges, universities, adult education centers, fitness studios, and various public places in Germany and Austria in November and December 2017. Participation was voluntary and without reimbursement. Each participant was given a piece of paper informing them about the voluntary and anonymous nature of their participation, the purpose of the study, and the data to be collected. Upon providing consent, participants received one of two paper-and-pencil questionnaire versions: one using the two-category and the other using the six-category response scale. The two versions were alternated as questionnaires were handed out to participants. When a participant returned the questionnaire, the facilitator immediately checked for missing values and, if any were found, politely asked the participant to complete the missing questions. With this procedure, we achieved complete responses for all questionnaires.

No power analysis was performed to determine the sample size of the main study. Instead, the goal was to collect at least 500 subjects per response format condition over a period of six weeks. No data was analyzed before data collection ended. The final sample size of our main study is typical of those used in current validation studies that publish diagnostic results. According to Goretzko et al. ([2024](#ref-goretzkoEvaluatingModelFit2024)), the sample sizes used in typical validation studies in the journal Psychological Assessment from 2015 to 2020 were as follows: about 38 percent used sample sizes of 401 to 1000, and 25 percent sample sizes larger than 1000.

## Summary

The goal of this study was to evaluate newly constructed psychological scales using gold-standard items. To achieve this, we focused on the specific setting of physical traits (i.e., body height, body weight, and chronological age), where theoretically robust gold-standard items are available in the form of numerical self-reports. Using confirmatory factor analysis, we found that, contrary to our assumptions, even in this ideal setting, gold-standard items did not exhibit the highest loadings on their corresponding latent variables for all physical traits. After controlling for the gold-standard item and gender, we observed a residual psychological component for all traits, but the interpretation of this latent variable was not straightforward. The correlations between the latent variables also did not closely match those between the gold-standard items. All results were highly similar across questionnaire versions with two and six response categories, although reliability estimates were slightly lower for the two-category versions.

## Gold-Standard Items for Physical Traits

Despite the favorable setting when constructing psychological scales for physical traits and our carefully curated scale-construction process, gold-standard items did not fully function as expected with respect to our theoretical assumptions. Only on one of six scales (three physical traits measured via two response-scale versions each) did the gold-standard item exhibit a higher loading on the latent variable than the psychological items. That was on the six-category scale for body height, but not on the two-category version. Considering both the size and the measurement uncertainty of the factor loadings, it seems evident that the gold-standard items for weight and age loaded substantially lower on the latent variable than the best items of the respective scales. This finding raises questions about whether our psychological scales for weight and age accurately measure the physical traits they were designed to assess. Given this ambiguous construct validity, we would have expected the psychological items to capture a latent component beyond the physical trait itself. To examine this possibility, we conducted follow-up analyses controlling item responses for the gold-standard item and gender. However, the resulting residual latent variables proved difficult to interpret, both in terms of whether such a latent variable was present at all and how it could be meaningfully characterized (to further probe this issue, we also inspected the largest correlated errors in the one-factor models reported in <a href="#tbl-5" class="quarto-xref">Table 5</a>, but without gaining additional insights). This lack of clarity is consistent with prior work suggesting that psychological questionnaires targeting physical attributes do not necessarily reflect a well-defined psychological construct that guides behavior ([Hilbert et al., 2022](#ref-hilbertWhatMeasureEmpirical2022); [Van der Linden, 2016](#ref-vanderlindenUnidimensionalLogisticResponse2016)). A more detailed inspection revealed that this pattern varied across traits, but did not yield a coherent interpretation overall. For age, we found some indication that the residual latent variable may reflect cognitive or psychological age ([Barak & Schiffman, 1981](#ref-barakCognitiveAgeNonchronological1981)), as the highest-loading items all referred to mental decline. However, many age items showed negligible loadings, and the items tended to be either well predicted by the physical age item or strongly related to the residual latent variable, but rarely both. For weight, in contrast, items with high loadings on the residual factor were also substantially predicted by the physical item, suggesting no clear separation between physical and psychological components. For height, all items were well predicted by the physical item and showed small, relatively uniform residual loadings, providing some evidence for an additional latent dimension. However, the loading pattern did not indicate how this dimension could be interpreted. Taken together, these findings provide no consistent support for a meaningful psychological latent variable beyond the assessed physical traits of body height and weight. This interpretation is further reinforced by the mismatch between the correlations among the latent variables and those among the corresponding gold-standard items.[1] Overall, only the height scales performed close to expectations, and we therefore consider them the only appropriate measures of their respective physical trait.

In light of these issues, we sought to rule out the possibility that certain decisions in our questionnaire construction process introduced idiosyncratic problems[2] by comparing our results with those of Bortolotti et al. ([2013](#ref-bortolottiRelevanceAdvantagesUsing2013)), who applied a similar procedure with 27 items and two response categories to examine a gold-standard item for body height. They found a correlation of .86 between the estimated person parameter from a two-parameter logistic model and self-reported height, which closely matches our correlation between the gold-standard height item and the latent variable (two response categories: .87, six response categories: .86). The comparison does not suggest that our scales have unique problems, as we were able to conceptually replicate Bortolotti et al. ([2013](#ref-bortolottiRelevanceAdvantagesUsing2013)) with our independently developed questionnaire.

It is also noteworthy that our results were surprisingly robust across different response formats, with very similar findings for questionnaire versions with two and six response categories. We used a six-category response format to align with literature advising against the use of a middle category ([Bradley et al., 2015](#ref-bradleyRatingScalesSurvey2015); [Masuda et al., 2017](#ref-masudaRespondentsLowMotivation2017)), and a two-category format because it is commonly used in symptom questionnaires within clinical psychology and medicine ([Clark & Watson, 2019](#ref-clarkConstructingValidityNew2019); [Rosellini & Brown, 2021](#ref-roselliniDevelopingValidatingClinical2021)). As expected, reliability estimates were slightly higher for scales with six response categories than for those with two ([Preston & Colman, 2000](#ref-prestonOptimalNumberResponse2000)). However, the results for model fit and validity were mixed. Some fit indices suggested a slightly better fit for items with two response categories and some for six categories (although the differences were rather small). The same holds true for the differences between the correlations among the gold-standard items and the correlations among the latent variables. Except for reliability, our results do not clearly favor either two or six response categories. While we would not expect substantially different results for scales with five or seven response categories, we cannot generalize our results to response formats that include a middle category (we only used formats without a middle category) or to response formats where all categories are labeled (we only labeled the endpoints). Hilbert et al. ([2016](#ref-hilbertInfluenceResponseFormat2016), [2022](#ref-hilbertWhatMeasureEmpirical2022)) showed that the same items, when measured with different numbers of response categories, do not necessarily measure the same latent variable. Thus, the selection of the response format may affect validity, prompting us to advise questionnaire developers to carefully consider the number of response categories when constructing psychological scales. An intensive cognitive pretesting phase can help clarify how respondents use and understand the categories, and empirical studies can help determine the number of response categories that provide better validity, reliability, and/or factor structures ([Willis & Artino, 2013](#ref-willisWhatDo2013)). In the current study, cognitive interviews and final results suggest that both applied response formats performed reasonably well.

## Gold-Standard Items for Psychological Traits

Based on our findings from using gold-standard items to evaluate psychological scales for physical traits, we want to discuss whether a similar strategy could also be used in the more common task of constructing psychological scales for psychological traits, where theoretically strong gold-standard items are more difficult to identify.

In psychology, we try to use precise names for our latent variables, such as emotional intelligence, self-efficacy, conscientiousness, or narcissism. It might, therefore, seem straightforward to turn these construct labels into corresponding “gold-standard” items. But is it reasonable to expect that, in a questionnaire that measures narcissism, an item such as *“I am a narcissist.”* (see [Konrath, 2014](#ref-konrathDevelopmentAnd2014)) most accurately reflects the construct and, consequently, shows the highest factor loading? This question is closely related to psychological research on single-item measures, that is, the measurement of psychological constructs using a single question in a rating-scale format, which often uses slightly longer wording and provides additional context in its instructions. Single-item measures are increasingly used in psychological research when constructs are narrowly defined and when there are strong practical reasons that longer multi-item scales cannot be applied without problems ([Fuchs & Diamantopoulos, 2009](#ref-fuchsUsingSingle2009)). However, rigorous validation is repeatedly emphasized as equally important for single-item measures as for all psychological scales ([Allen et al., 2022](#ref-allenSingleItem2022)). In this literature, a common approach to assessing the validity of a newly constructed single-item measure is to examine its loading in a one-factor model that includes both the single-item measure and the items from an established multi-item scale for the same psychological construct ([Fisher et al., 2016](#ref-fisherDevelopingInvestigating2016); [Matthews et al., 2022](#ref-matthewsNormalizingUse2022)). This is exactly the same logic we used in our study of gold-standard items for physical traits.

Thus, when constructing a new psychological scale, researchers could check whether a validated single-item measure of the same psychological construct is available and, if so, use it as a natural candidate for a gold-standard item in their validation study. Although it is argued that a good single-item measure should load highly on the latent variable shared with the multi-item scale, it is not required for the single-item measure to have a loading equal to or higher than those of the best items in the scale ([Fisher et al., 2016](#ref-fisherDevelopingInvestigating2016); [Matthews et al., 2022](#ref-matthewsNormalizingUse2022)). We would argue that in most circumstances, it would nonetheless suggest issues with the newly constructed scale if the single-item measure has a low loading compared to the other items. Empirical evidence suggests that such analyses can detect problems in a substantial number of scale-construction processes. In Fisher et al. ([2016](#ref-fisherDevelopingInvestigating2016)), eight out of 18 newly constructed single-item measures had lower loadings than all items of the corresponding multi-item scale, and one single-item measure had a low absolute loading under 0.3. In Matthews et al. ([2022](#ref-matthewsNormalizingUse2022)), seven out of 91 newly developed single-item measures did not even load significantly on the factor shared with the corresponding multi-item scale, loaded in the wrong direction, or had a low absolute loading under 0.3.[3] Even if there is no consensus on how high the single-item measure’s loading should be ([Allen et al., 2022](#ref-allenSingleItem2022)), most experts would agree that, in the cases reported above, there must be a problem with the single-item measure, the multi-item scale, or both.

What plausible reasons might explain why a single-item measure does not exhibit the highest loading relative to items from a corresponding multi-item scale? Prior research suggests that single-item measures are most effective when the underlying psychological construct is narrow in scope and low in complexity ([Fuchs & Diamantopoulos, 2009](#ref-fuchsUsingSingle2009); [Matthews et al., 2022](#ref-matthewsNormalizingUse2022)). For such homogeneous constructs, it is comparatively easy to formulate an item that captures the construct’s core meaning and is interpreted consistently across respondents. From this perspective, our application of constructing psychological scales for physical traits such as height, weight, and age should have provided particularly favorable conditions. In contrast, more heterogeneous and complex constructs, such as personality traits, are inherently more difficult to capture with a single item without introducing ambiguity. Although it has been argued that highly abstract single-item measures may still be useful for such constructs, especially when it is impractical to cover each content facet with an item ([Fuchs & Diamantopoulos, 2009](#ref-fuchsUsingSingle2009)), any single-item measure in this context reflects a specific weighting of the construct’s multiple aspects. Consequently, a single-item measure can be expected to show the highest loading only if the aspect it emphasizes is also the dominant component in the corresponding multi-item scale. More generally, if a construct is sufficiently heterogeneous that no single item can be expected to dominate, it is questionable whether a unidimensional latent variable model is appropriate in the first place, as multiple factors and/or correlated errors are likely. In sum, we concur with prior literature that one should not assume that a single-item measure will necessarily exhibit the highest loading when combined with items from a corresponding multi-item scale. Instead, researchers should carefully compare the construct definitions underlying the multi-item scale and the single-item measure prior to validation. Such a theoretical analysis allows for the formulation of more informative expectations regarding the magnitude of the single-item loading than the minimal assumption that it merely differs from zero.

But what can be done in practice if no validated single-item measure closely matches the construct definition of a newly constructed scale? Can a gold-standard item still be defined in a broader sense? We argue that there are many settings in which this is both feasible and informative. Consider, for example, the measurement of personality using adjectives. Suppose we aim to assess the latent variable *interpersonal warmth* with items such as *affectionate*, *friendly*, *talkative*, *unprejudiced*, each rated on a scale from *not at all* (1) to *fully and entirely* (6) with six response categories. If these items indeed capture interpersonal warmth, they should also correlate with the adjective *warm* measured using the same response format. More importantly, one might question the construct definition itself if this straightforward item does not exhibit the highest loading in a one-factor model when included alongside the other items. In such cases, it is plausible that a different construct is being measured, for instance, *friendliness*, if the item *friendly* shows the highest loading. In this sense, the item *warm* can serve as a natural gold-standard item, analogous to the role of gold-standard indicators in our example involving physical traits. This raises the further question of whether such an item should be included in the final scale. We would argue that this decision depends on the specific research context and cannot be resolved in general terms. If the scale was originally intended to include an item directly reflecting the construct label (e.g., warm), its inclusion would be conceptually straightforward. However, even if this were not the case, the structure of adjective-based questionnaires makes it difficult to justify omitting the most direct representation of the construct from empirical consideration. Ultimately, the decision of whether such items are retained as part of the final scale or used solely for validation purposes remains a matter of researcher judgment. Future studies should investigate different strategies of finding gold-standard items in more detail. For example, we are interested in using digital behavioral indicators from mobile sensing ([Schoedel et al., 2026](#ref-schoedelFromDigital2026); [Schoedel & Mehl, 2024](#ref-schoedelMobileSensing2024)) to validate scales with behavior-based definitions like some personality facets.

Despite these examples, we acknowledge that there will be many instances in which new psychological scales are constructed when theoretically strong gold-standard items are not available, and our suggested evaluation strategy cannot be readily applied. This should not be a problem, as it does not, in theory, invalidate the settings already discussed in which the strategy is meaningful. It is important to note that our evaluation strategy using gold-standard items is only a special case of a more general principle in test construction: Whenever constructing a new psychological scale, researchers should use their theoretical knowledge of the intended construct, its nomological network, and the scale’s intended practical use cases to generate hypotheses about the scale and its items that are at least partly open to empirical testing. The more hypotheses about a new scale are available and the stronger the theoretical rationale for expecting a certain pattern, the more holistically researchers can empirically evaluate whether they actually measure the intended construct, and the better they should be able to convince the scientific community that they have constructed a useful new scale. Assessing the loadings of self-postulated gold-standard items is just one small example of such a hypothesis, and how convincing that hypothesis is will always depend on the concrete example.

Given that we did not consistently manage to construct scales in which gold-standard items had the highest factor loadings, even under the optimal setting of physical traits, we expect it to be even more difficult to achieve the theoretically desirable loading pattern when constructing scales for psychological traits. Nevertheless, we still think that gold-standard items can be a very useful tool for validating new psychological scales.

## General Lessons and Recommendations

Despite following a rigorous, best-practice questionnaire construction process, our inability to construct valid psychological scales for physical traits that fully met our self-imposed criteria adds to a growing body of work highlighting the limitations of standard scale-construction approaches ([Goretzko et al., 2020](#ref-goretzkoNotVeryPowerful2020); [Hilbert et al., 2016](#ref-hilbertInfluenceResponseFormat2016), [2022](#ref-hilbertWhatMeasureEmpirical2022); [Pargent et al., 2019](#ref-pargentCanMakeIt2019)). Taken together, these findings call for critical reflection on common practices in item construction. We summarize our main insights below.

First, the choice of an optimal response format appears less clear-cut than often assumed (e.g., [Lee & Paek, 2014](#ref-leeSearchOptimalNumber2014)). In the present study, items with only two response categories were not clearly inferior, despite common concerns in the literature. Although our binary scales showed lower reliability (as expected, based on [Preston & Colman, 2000](#ref-prestonOptimalNumberResponse2000); [Revilla et al., 2014](#ref-revillaChoosingNumberCategories2014)), we observed no consistent disadvantages in terms of validity or model fit. This highlights that reliability reflects measurement precision, not validity. This does not necessarily contradict attempts to select items based on reliability (e.g., see [Zijlmans et al., 2019](#ref-zijlmansItemScoreReliabilitySelection2019)) since items must have high loadings to achieve high reliability estimates for the scale. However, because reliability is easier to control for by adding more suitable items, construct validity seems to be the stronger argument when choosing the number of response categories. We therefore argue that decisions about response formats should be guided primarily by considerations of validity. In practice, this requires empirical comparison of alternative response formats, especially in high-stakes contexts such as clinical assessment or personnel selection. Importantly, the number of response categories may affect the latent variable being measured ([Hilbert et al., 2016](#ref-hilbertInfluenceResponseFormat2016), [2022](#ref-hilbertWhatMeasureEmpirical2022)), making such comparisons essential.

At the same time, all items should undergo thorough cognitive pretesting including the response format. Cognitive interviews or standardized pretesting frameworks ([Gehlbach & Brinkworth, 2011](#ref-gehlbachMeasureTwice2011); [Wolf et al., 2025](#ref-wolfResponseProcessEvaluation2025)) can help ensure that respondents understand item wording and response options as intended, identify categories that are systematically overlooked, and assess the cognitive demands imposed by the questionnaire ([Tourangeau et al., 2000](#ref-tourangeauPsychologySurvey2000)). This is particularly relevant given that simpler formats, such as binary scales, may reduce respondent burden ([Hilbert et al., 2016](#ref-hilbertInfluenceResponseFormat2016)).

Second, overall model fit should not be overinterpreted as evidence of item quality. Good model fit does not guarantee that items are well formulated or substantively meaningful. Even when best-practice guidelines are followed during item construction, the resulting models may not outperform those in which items were purposely constructed in violation of such rules ([Goretzko et al., 2020](#ref-goretzkoNotVeryPowerful2020); [Pargent et al., 2019](#ref-pargentCanMakeIt2019)). In the present study, even the a priori most suitable gold-standard items did not consistently exhibit the highest loadings. Thus, relying solely on statistical criteria, such as factor loadings, to evaluate items is insufficient. Instead, item selection should always consider both statistical performance and substantive content.

Finally, we emphasize the central importance of content validity in scale development. Using gold-standard items (i.e., items for which high loadings are theoretically expected) represents a special case of a broader strategy: formulating explicit hypotheses about the role and behavior of individual items in a newly constructed scale. Such hypotheses may concern, for example, the ranking of factor loadings or the pattern of correlations with construct-divergent measures. Incorporating these expectations into the validation process can yield stronger, more informative tests of whether a scale adequately captures the intended latent variable.

# CFA models with correlated errors

[1] One reviewer raised the question whether measurement error in the self-reported physical items could explain the reported mismatch in the correlation pattern. In our adult population, we would expect reporting error to be negligible for age and small for height, with greater potential for reporting inaccuracies in weight. Such error could contribute to some discrepancies. However, this does not explain the pattern that the observed correlation between weight and height was substantially smaller for the latent variables than for the physical items. If the physical weight items were strongly affected by measurement error, the true correlation between the physical traits should be even higher than the correlation we observed between the physical items.

[2] For example, one reviewer questioned why many of our items use quite extreme item wording (e.g., “I have already lived most of my life”) and pondered whether this might explain why our scales did not work as expected. However, our cognitive interviews and pilot study suggested that – maybe surprisingly – these items worked best to assess height, weight or age.
